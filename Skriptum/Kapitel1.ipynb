# Kapitel 1 - Digitale Grundkompetenzen

Hier werden wir uns mit grundlegenden digitalen Kompetenzen beschäftigen, wie zum Beispiel Programmieren.
Bevor wir das aber tun, lohnt sich ein kurzer Blick auf etwas Größeres:
auf digitale Ungleichheit – oft als Digital Divide bezeichnet.

## Digital Divide – was heißt das?

Mit **Digital Divide** meint man die Gräben zwischen Menschen und Gruppen, wenn es um digitale Themen geht:
* Wer hat einen halbwegs aktuellen Computer, ein Smartphone, eine stabile Internetverbindung?
* Wer hat Zugang zu guter Bildung, Kursen, Zeit und Ruhe, um sich mit solchen Dingen zu beschäftigen?
* Wer hat Menschen in seinem Umfeld, die helfen können, wenn etwas nicht funktioniert?

Diese Unterschiede sind real – weltweit, aber auch innerhalb eines Landes, einer Stadt, einer Klasse.
Sie betreffen nicht nur „Komfort“, sondern ganz grundlegende Fragen:
Wer kann an moderner Arbeit, Bildung und Gesellschaft **teilhaben** – und wer nicht?

In diesem Kapitel gehen wir stillschweigend davon aus, dass du:

* einen Computer und Internetzugang hast,
* Software installieren darfst,
* und dir überhaupt Zeit nehmen kannst, dich mit Programmieren zu beschäftigen.

Das ist leider nicht selbstverständlich. Wir sollten es also **wertschätzen**.

## Privileg und Verantwortung

Wenn du **digital affin** bist, oder es in diesem Kurs werden kannst, bedeutet das zweierlei:

Du hast Chancen, die andere nicht haben
Du kannst lernen, wie Computer funktionieren, wie man Daten auswertet, Modelle baut und Programme schreibt.
Das eröffnet dir später berufliche und wissenschaftliche Möglichkeiten.

Du hast eine Verantwortung

* im eigenen **Handeln**: Technik so zu nutzen, dass sie nicht ausschließt, sondern hilft;
* im **Umgang mit anderen**: Wissen weiterzugeben, verständlich zu bleiben, statt mit Fachsprache Mauern aufzubauen;
* in **Projekten**: mitzudenken, wer durch digitale Lösungen mitgemeint ist – und wer gerade übergangen wird.

Dieses Kapitel ist deshalb zweifach gemeint:

Es soll dir ganz praktisch zeigen, wie man mit Python und VS Code arbeitet.
Es soll dich gleichzeitig daran erinnern, dass der Umgang mit digitaler Technik nicht neutral ist,
sondern immer auch mit Fragen von Zugang, Macht und Gerechtigkeit zu tun hat.

Mit diesem Blick im Hinterkopf beginnen wir nun mit den technischen Grundlagen:
Was Programmieren ist, was Python ist, wie du Python installierst – und wie du Schritt für Schritt deine ersten eigenen Programme schreibst.

## Was ist Programmieren?

Unter **Programmieren** versteht man ganz allgemein die Tätigkeit, einem Computer präzise Anweisungen zu geben.
Immer wenn du einem Gerät erklärst, was es Schritt für Schritt tun soll, programmierst du – egal ob das ein Taschenrechner, ein Laptop oder ein großes Rechencluster ist.

Ein Taschenrechner kennt nur wenige Befehle: plus, minus, mal, geteilt, vielleicht noch ein paar Funktionen wie sin oder sqrt.
Moderne Computer können viel mehr: Simulationen laufen lassen, große Datensätze analysieren, Grafiken zeichnen, Text verarbeiten, mit dem Internet kommunizieren – aber nur, wenn wir ihnen sagen wie. Genau hier kommt Programmieren ins Spiel:

Wir beschreiben ein Problem (zum Beispiel: „Wie entwickeln sich die CO₂‑Emissionen eines Energiesystems über 30 Jahre?“).
Wir übersetzen diese Beschreibung in klare Befehle, die der Computer versteht.
Der Computer führt diese Befehle aus, viel schneller und zuverlässiger, als wir das mit Papier und Stift jemals könnten.

Programmieren ist dabei nicht nur ein technischer Vorgang. Es schult auch:


* deine Fähigkeit, **komplexe Probleme** in kleinere Schritte zu zerlegen,
* deine **Genauigkeit** im wissenschaftlichen Arbeiten,
* und deine **Kreativität**, Modelle zu bauen, zu testen und zu verbessern.

## Was ist Python?

In dieser Lehrveranstaltung – und in diesem Skript – verwenden wir die **Programmiersprache Python**.

Python ist aus mehreren Gründen eine gute Wahl:

* Einfach und lesbar: Die Grammatik (die sogenannte Syntax) ist vergleichsweise klar und wenig überladen. 
* Weit verbreitet in der Wissenschaft: In Natur‑ und Sozialwissenschaften, Klimaforschung, Datenanalyse, Machine Learning usw. 
* Kostenlos und offen: Python ist frei verfügbar, du brauchst keine Lizenz zu kaufen.
* Viele Zusatzmodule: Es gibt Bibliotheken für fast alles: von Klimadaten über Netzwerkanalyse bis hin zu statistischen Modellen.
* Große Community: Wenn du ein Problem hast, ist die Wahrscheinlichkeit hoch, dass jemand es schon hatte und im Internet eine Lösung beschrieben hat.

### Was ist VS Code – und warum verwenden wir es?

Um Python‑Programme zu schreiben, brauchst du zwei Dinge:

* die Programmiersprache selbst (Python),
* eine Umgebung, in der du bequem Code schreiben, speichern und ausführen kannst.

Eine solche Umgebung nennt man häufig Editor oder IDE (Integrated Development Environment).
In diesem Skript verwenden wir Visual Studio Code – kurz VS Code.

VS Code ist:
* ein Texteditor, in dem du Programmcode schreibst (also .py‑Dateien),
* gleichzeitig eine Entwicklungsumgebung: du kannst Python direkt aus VS Code starten, Fehlermeldungen sehen, Dateien organisieren, Erweiterungen installieren,
* frei verfügbar für Windows, macOS und Linux.

Du kannst dir VS Code ungefähr wie eine sehr leistungsfähige Version von „Word“ vorstellen, die speziell darauf ausgelegt ist, dir beim Programmieren zu helfen:

* Schlüsselwörter werden farbig hervorgehoben,
* du bekommst Vorschläge und Autovervollständigung,
* du kannst mit wenigen Klicks ein Programm starten und das Ergebnis sehen.

In den folgenden Abschnitten dieses Kapitels richten wir VS Code und Python so ein, dass du sofort loslegen kannst.



## Installation von VS Code

Im ersten Schritt installieren wir VS Code auf deinem Computer.
Die folgenden Hinweise beziehen sich beispielhaft auf Windows, sind aber auf macOS und Linux sehr ähnlich.

**Webseite öffnen**

Öffne deinen Browser und gehe auf https://code.visualstudio.com

**Version auswählen und herunterladen**

Auf der Startseite wird dir normalerweise automatisch eine passende Version für dein Betriebssystem vorgeschlagen (z.B. „Download for Windows“).
Klicke auf den Download‑Button und warte, bis die Installationsdatei heruntergeladen ist.

**Installationsdatei starten**

Suche die heruntergeladene Datei (unter Windows meist im Ordner Downloads) und doppelklicke sie.

**Installationsassistent durchklicken**

Akzeptiere die Lizenzbedingungen, wähle einen Installationsort (die Voreinstellung ist meist sinnvoll),optional kannst du VS Code im Startmenü und/oder auf dem Desktop verknüpfen lassen.Es ist in Ordnung, hier einfach die vorgeschlagenen Einstellungen zu übernehmen.

**Installation abschließen**

Klicke auf „Install“ und danach auf „Finish“.
VS Code ist nun auf deinem System installiert. Du findest es im Startmenü (Windows) oder im Programme‑Ordner (macOS).
Wenn du VS Code das erste Mal startest, siehst du eine leere Oberfläche mit einem Startbildschirm.
Damit VS Code mit Python zusammenarbeitet, werden wir im nächsten Schritt die Python‑Erweiterung installieren und eine erste kleine Datei anlegen – genau das achen wir jetzt.

## Python installieren (damit VS Code etwas ausführen kann)

VS Code allein reicht noch nicht: Er ist der Ort, an dem du den Code schreibst, aber ausführen kann er ihn nur, wenn auf deinem Computer auch Python installiert ist.

Da die meisten von euch Python noch nicht installiert haben werden, beginnen wir mit der Installation.


Öffne deinen Browser und gehe auf: https://www.python.org/downloads
Oben auf der Seite wird dir eine aktuelle Version von Python 3 angeboten (z.B. „Download Python 3.11.x“).
Diese Version ist für unseren Kurs völlig ausreichend – du kannst einfach die vorgeschlagene nehmen.
Klicke auf den Download‑Button und warte, bis die Installationsdatei heruntergeladen ist.
Doppelklicke auf die heruntergeladene Datei, um die Installation zu starten.
Damit ist Python installiert.
Jetzt wollen wir sicherstellen, dass dein System Python auch wirklich erkennt und ausführen kann.



### Installation überprüfen

Es gibt viele Wege, zu versuchen, ob Python „funktioniert“:

* du könntest in VS Code irgendwelche Einstellungen öffnen,
* in den Systemeinstellungen nachschauen,
* nach Icons suchen, die vielleicht „Python“ heißen.

All das ist eher umständlich und liefert oft keine klare Antwort.
Der einfachste und direkteste Weg ist:
Wir reden per Textbefehl mit dem System und fragen:
„Welche Python‑Version kennst du?“
Dafür verwenden wir das **Terminal**.
Schauen wir uns kurz an, was das ist.



### Was ist ein Terminal?

Du kennst deinen Computer wahrscheinlich vor allem so:

* Du klickst mit der Maus auf Ordner und Dateien.
* Du startest Programme durch Doppelklick.
* Du siehst Fenster, Buttons und Menüs.

Das ist die grafische Oberfläche deines Betriebssystems.
Parallel dazu gibt es noch eine zweite Art, den Computer zu bedienen: eine sehr viel mächtigere:
Über **Textbefehle**. Ein **Terminal** (unter Windows oft „Eingabeaufforderung“ oder „Command Prompt“) ist ein Fenster, in das direkt Befehle an deinen Computer senden kannst. Somit kannst du nicht nur jedes Programm auf deinem Computer starten, sondern zum Beispiel auch Einstellungen ändern oder Befehle ausführen, die man auf der grafischen Oberfläche gar nicht sehen kann.

Für uns ist das Terminal wichtig, weil wir dort sehr direkt prüfen können ob deine Python-Installation funktioniert.



### Terminal öffnen

Je nach Betriebssystem machst du das so:

Windows:
Öffne das Startmenü und gib cmd ein.
Klicke auf „Eingabeaufforderung“ oder „Command Prompt“.
Alternativ kannst du auch „Windows‑Terminal“ verwenden.

macOS:
Öffne Spotlight (Lupe oben rechts oder Cmd+Leertaste).
Gib Terminal ein und starte das Programm mit diesem Namen.

Linux:
Öffne das Anwendungsmenü und suche nach „Terminal“, „Konsole“ oder ähnlichem.


Wenn das Terminal geöffnet ist, siehst du eine Zeile, die in etwa so aussieht:

`C:\Users\DeinName>`

oder

`deinname@rechner:~$`

Das ist die Stelle, an der du Befehle eingeben kannst.



### Prüfen, ob Python richtig installiert ist

Jetzt testen wir: 

Tippe im Terminal: `python --version` und drücke Enter. 

Wenn das nicht funktioniert (Fehlermeldung), versuche:
`python3 --version` 

Nun sollte eine Versionsnummer erscheinen, z.B.:

`Python 3.11.5`

Perfekt!
Python ist installiert und dein System kennt den Befehl python (bzw. python3).
Das heißt, du kannst später im Terminal und in VS Code Python‑Programme ausführen.
Wir können nun also VS Code mit Python verbinden und unser erstes kleines Programm schreiben.

## VS Code mit Python verbinden

Jetzt haben wir die beiden Bausteine:

* VS Code zum Schreiben,
* Python zum Ausführen.

Im letzten Vorbereitungsschritt sorgen wir dafür, dass VS Code Python gut versteht.

### Python-Erweiterung in VS Code installieren

Starte VS Code. Du siehst links eine Seitenleiste mit Symbolen. Eines davon sieht aus wie vier kleine Quadrate, die ein größeres Quadrat bilden – das ist der Bereich Extensions (Erweiterungen).
Klicke links auf Extensions.
Gib oben im Suchfeld Python ein.
Wähle die Erweiterung „Python“ von Microsoft aus (sie ist meist ganz oben).
Klicke auf Install.
Diese Erweiterung sorgt u.a. dafür, dass VS Code:

* Python‑Code erkennt und farbig hervorhebt,
* dir beim Schreiben hilft (Autovervollständigung),
* und dir einfache Möglichkeiten bietet, ein Python‑Programm zu starten.

Damit ist unsere Grundumgebung fertig vorbereitet.



## Dein erstes Programm

Um das Ganze greifbar zu machen, schreiben wir jetzt unser allererstes Programm.
Es wird keine Klimamodelle lösen, keine großen Datensätze analysieren – es wird einfach nur eine einfache Rechnung lösen.
Das klingt unspektakulär, ist aber der erste vollständige Weg:

Datei anlegen → Code schreiben → Programm starten → Ergebnis sehen.

### Eine neue Datei in VS Code anlegen

Klicke in VS Code links oben auf File → New File (oder verwende Ctrl+N / Cmd+N).
Speichere die Datei gleich ab:
File → Save As…
Gib ihr einen Namen, z.B. erstes_programm.py
Wähle einen Ordner, den du leicht wiederfindest (z.B. einen neuen Ordner in deinen Dokumenten).
Wichtig: Die Datei sollte auf .py enden – daran erkennt VS Code und Python, dass es sich um eine Python‑Datei handelt.

### Code schreiben

In die leere Datei schreibst du jetzt:

```
result = 80 / 8300
print(result)
```

Was passiert hier?
`result = 80 / 8300` weist der Variable `result` das Ergebnis der Rechenoperation 80 dividiert durch 8300 zu.
`print(result)` sorgt dafür, dass der Computer den Wert von zahl auf dem Bildschirm ausgibt.

Noch ist das sehr einfach – genau so soll es für den Einstieg sein.
Wir wollen zuerst nur die Mechanik üben: wie du einen kleinen Code schreibst, speicherst und ausführst.

Das sind übrigens nicht irgendwelche Zahlen. Ich hab ein wenig gekürzt, damit die Zahlen nicht so lang sind. Die untere erkennt ihr aber bestimmt: 8.3 Milliarden. So viele Menschen gibt es etwa auf der Erde. Und die Zahl darüber: 80 Millionen. So viele Menschen auf der Erde haben in etwa schon einmal in ihrem Leben Computercode geschrieben. Weniger als 0.1%. Zu denen gehört ihr jetzt. Das impliziert auch eine gewisse Verantwortung, die wir im Laufe des Semesters genauer erkunden werden. Aber fürs erste, starten wir einfach unser Programm.

### Programm aus VS Code heraus starten

Es gibt verschiedene Wege, ein Python‑Programm aus VS Code zu **starten**.
Für den Anfang reicht eine sehr einfache Variante:
Stelle sicher, dass deine Datei (erstes_programm.py) gespeichert ist.
Je nach VS‑Code‑Setup siehst du oben rechts in der Editor‑Leiste ein kleines „Play“-Symbol (Dreieck) mit „Run Python File“.
Wenn du dieses Symbol anklickst, versucht VS Code, die Datei mit Python auszuführen.
Alternativ kannst du das integrierte Terminal verwenden:
Öffne das Terminal mit View → Terminal.
Du siehst unten ein neues Fenster mit einer Eingabezeile.
Dort gibst du ein:
python erstes_programm.py
(oder python3, je nach Installation)
Wenn alles richtig eingerichtet ist, sollte im Terminal folgendes erscheinen:

`0.00963855421686747`

Damit hast du dein erstes Programm geschrieben und ausgeführt.

![Success](https://upload.wikimedia.org/wikipedia/en/f/ff/SuccessKid.jpg)

### Warum dieses Programm wichtig ist

Natürlich wird dich ein Programm, das nur eine Division ausführt, fachlich nicht lange begeistern.
Der Punkt ist ein anderer: Du hast jetzt mindestens einmal erlebt,

* wie du eine Python‑Datei anlegst,
* wie du Code schreibst,
* wie VS Code dir dabei hilft,
* und wie diesen Code ausführst.

Ab hier können wir Schritt für Schritt mehr Inhalt einbauen:

Das Muster bleibt aber gleich:
Du formulierst ein Problem.
Du übersetzt es in Code.
Der Computer führt den Code aus.

In den nächsten Kapiteln werden wir genau diesen Ablauf verwenden, um nach und nach zu immer spannendere Fragen zu kommen – bis hin zu Klimapolitik und sozialer Gerechtigkeit im letzten Kapitel.

## GitHub

Neben **Programmieren** und dem Benutzen eines **Terminals** gehört noch ein weiterer Punkt zu digitalen Grundkompetenzen.
Es geht um die Frage, was man mit Code macht, wenn man ihn dann gecoded hat. Wie kann man ihn teilen, verbessern und gemeinsam etwas bewirken. Dazu werden wir **GitHub** kennen lernen.
Wir werden es verwenden, um dieses Skriptum und Beispielprogramme herunterzuladen.

### Was ist GitHub?

GitHub ist so etwas wie:

* eine große Bibliothek für Programmier‑Projekte,
* kombiniert mit einer Plattform zum gemeinsamen Arbeiten.

Menschen und Organisationen laden dort ihren Code und ihre Materialien hoch,
damit andere sie:

* lesen,
* herunterladen,
* verbessern oder
* weiterverwenden können.

Viele bekannte Open‑Source‑Projekte liegen auf GitHub – also Software, die frei verfügbar ist.

### Wie benutzt du GitHub für dieses Skriptum?

Für unseren Kurs brauchst du nur die einfache Variante:
Du bekommst einen Link zu einem GitHub‑Projekt (Repository), in dem:
* das Skriptum,
* und teils dazugehörige Python‑Dateien liegen.

Wenn du diesen Link im Browser öffnest, siehst du:
* eine Liste von Dateien,
* Ordner,
* und oft schon eine Vorschau (z.B. von PDF‑ oder Textdateien).

Um alles auf einmal herunterzuladen:
* Klicke auf den grünen Button „Code“ (oder ähnlich),
* wähle dort „Download ZIP“ aus,
* speichere die ZIP‑Datei und
* entpacke sie in einen Ordner auf deinem Computer.

Danach hast du das komplette Skriptum und alle Beispielprogramme lokal –
du kannst sie in VS Code öffnen, bearbeiten und ausführen, ohne wieder zu GitHub zu müssen.

Unser GitHub Repository findest du unter
[https://github.com/georgjaeger/computational-basics](https://github.com/georgjaeger/computational-basics)

### Was kann GitHub, wenn man es besser beherrscht?

Wenn man sich mit GitHub genauer beschäftigt, kann man damit:

**Versionsverwaltung betreiben**
– jede Änderung wird gespeichert, man kann jederzeit zu älteren Ständen zurückspringen.

**Gemeinsam an Code arbeiten**
– mehrere Personen entwickeln gleichzeitig am selben Projekt, ohne sich gegenseitig zu stören.

**Fehler melden und beheben**
– über sogenannte „Issues“ kann man Probleme dokumentieren und nachverfolgen.

**eigene Projekte veröffentlichen**
– damit andere sie benutzen, kommentieren oder weiterentwickeln können.

Wie das technisch genau funktioniert (z.B. mit „Commits“, „Branches“ oder „Pull Requests“), müssen wir an dieser Stelle noch nicht wissen. Für uns ist es nur eine alternative Variante an dieses Skriptum zu kommen. Trotzdem ist das Wissen um das Potential von GitHub sehr wichtig.
